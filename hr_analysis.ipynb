```python
numeric_data = df.select_dtypes(
    include=np.number
)

correlation = numeric_data.corr()

print("\n===== CORRELATION MATRIX =====")

print(correlation)
```

```python
correlation_pairs = (
    correlation
    .where(
        np.triu(
            np.ones(correlation.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(
        ascending=False
    )
)

print("\n===== STRONGEST POSITIVE CORRELATIONS =====")

print(correlation_pairs.head(10))
```

```python
# ==========================================
# STEP 10: STATISTICAL ANALYSIS
# ==========================================

print("\n===== STATISTICAL ANALYSIS =====")

# Separate salary values by attrition
salary_stayed = df.loc[
    df["Attrition"] == "No",
    "MonthlyIncome"
]

salary_left = df.loc[
    df["Attrition"] == "Yes",
    "MonthlyIncome"
]

print("Stayed employees:", len(salary_stayed))
print("Left employees:", len(salary_left))
```

```python
satisfaction_stayed = df.loc[
    df["Attrition"] == "No",
    "JobSatisfaction"
]

satisfaction_left = df.loc[
    df["Attrition"] == "Yes",
    "JobSatisfaction"
]

t_stat_satisfaction, p_value_satisfaction = stats.ttest_ind(
    satisfaction_stayed,
    satisfaction_left,
    equal_var=False
)

print("\n===== JOB SATISFACTION T-TEST =====")

print("T-statistic:", t_stat_satisfaction)
print("P-value:", p_value_satisfaction)
```

```python
department_table = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

print("\n===== DEPARTMENT VS ATTRITION =====")
print(department_table)

chi2, p_value, dof, expected = stats.chi2_contingency(
    department_table
)

print("\nChi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
```

```python
# ==========================================
# STEP 11: BUSINESS INSIGHTS
# ==========================================

print("\n===== BUSINESS INSIGHTS =====")

total_employees = len(df)

employees_left = (df["Attrition"] == "Yes").sum()

attrition_rate = employees_left / total_employees * 100

print("\n1. Overall Attrition Rate:")
print(round(attrition_rate, 2), "%")
```

```python
department_rate = (
    pd.crosstab(
        df["Department"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

highest_department = (
    department_rate["Yes"]
    .idxmax()
)

highest_department_rate = (
    department_rate["Yes"]
    .max()
)

print("\n2. Department with Highest Attrition:")
print(highest_department)

print("Attrition Rate:",
      round(highest_department_rate, 2), "%")
```

```python
jobrole_rate = (
    pd.crosstab(
        df["JobRole"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

highest_jobrole = (
    jobrole_rate["Yes"]
    .idxmax()
)

highest_jobrole_rate = (
    jobrole_rate["Yes"]
    .max()
)

print("\n3. Job Role with Highest Attrition:")
print(highest_jobrole)

print("Attrition Rate:",
      round(highest_jobrole_rate, 2), "%")
```

```python
salary_stayed_mean = df.loc[
    df["Attrition"] == "No",
    "MonthlyIncome"
].mean()

salary_left_mean = df.loc[
    df["Attrition"] == "Yes",
    "MonthlyIncome"
].mean()

salary_difference = (
    salary_stayed_mean - salary_left_mean
)

print("\n4. Salary Comparison:")

print("Average salary of employees who stayed:",
      round(salary_stayed_mean, 2))

print("Average salary of employees who left:",
      round(salary_left_mean, 2))

print("Difference:",
      round(salary_difference, 2))
```

```python
experience_stayed = df.loc[
    df["Attrition"] == "No",
    "YearsAtCompany"
].mean()

experience_left = df.loc[
    df["Attrition"] == "Yes",
    "YearsAtCompany"
].mean()

print("\n7. Years at Company:")

print("Stayed:",
      round(experience_stayed, 2))

print("Left:",
      round(experience_left, 2))
```
